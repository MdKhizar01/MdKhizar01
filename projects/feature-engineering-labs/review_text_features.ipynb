## Task 1.2 Create your own naive tokenization method (whitespace-based), and apply it to the text given in the task 1.1
note: 1. do not use the existing togkenization methods given by NLP; 2. split the words by whitespace character, the output is more likely as the unigram; 3. no repeating elements in the output.

```python
# write you code here
# Naive whitespace-based tokenization method
def naive_tokenizer(text):
    # Split the text by whitespace characters
    tokens = text.split()


    tokens = [token.lower() for token in tokens]


    unique_tokens = list(set(tokens))

    return unique_tokens

# Apply the naive tokenization method to the given text
train_text = """My wife took me here on my birthday for breakfast and it was excellent.  The weather was perfect which made sitting outside overlooking their grounds an absolute pleasure.
Our waitress was excellent and our food arrived quickly on the semi-busy Saturday morning.  It looked like the place fills up pretty quickly so the earlier you get here the better.

Do yourself a favor and get their Bloody Mary.  It was phenomenal and simply the best I\'ve ever had.
I'm pretty sure they only use ingredients from their garden and blend them fresh when you order it.  It was amazing.

While EVERYTHING on the menu looks excellent, I had the white truffle scrambled eggs vegetable skillet and it was tasty and delicious.
It came with 2 pieces of their griddled bread with was amazing and it absolutely made the meal complete.
It was the best "toast" I\'ve ever had.

Anyway, I can\'t wait to go back!"""

# Get the tokenized output
naive_tokens = naive_tokenizer(train_text)

# Print the result
print("Naively Tokenized Words (No Duplicates):")
print(naive_tokens)
```

## **Question 1**. Given a sentence "He likes cat". In unigram representation, it could be "He", "likes", "cat". In bigram representation, it could be "He likes", "likes cat". In trigram representation, it could be "He likes cat". Explain why the storage and computation cost increase with the growth of n in n-gram methods.

Answer to Q1: The computing expense and the cost of storage rise with 'n' in n-grams approaches due to the exponential increase of unique n-grams that result from a full text database as 'n' increases.

To demonstrate this let's think about unigrams (n=1) is a dictionary of single words. In bigrams (n=2), you have consecutive two-word pairs, in trigrams (n=3), you have three-word consecutive sets, and so on.

Let's consider the sentence "He likes cat".

Unigrams: ["He", "likes", "cat"] - 3 unique unigrams.
Bigrams: ["He likes", "likes cat"] - 2 unique bigrams.
Trigrams: ["He likes cat"] - 1 unique trigram.
This example does not show probablistic growth that well. However, you can imagine a longer piece of text. Mentioning n=1 (unigrams) does not have much impact on growth of combinations, but as described above more combinations can be observed when the sequence length ('n') increases. As a result, depending on the nature of the text, you can have:

Higher Storage Cost: the vocabularies of a bigram will substantially larger than a unigram, and a trigram will substantially larger than bigrams. The storage of such large vocabularies and their counts, weights etc. will consequently require more storage in memory and disk.
Higher Computation Cost:
Generation: the generation of all possible n-grams from a text will involve processing the text through multiple times or using more complex and time-consuming algorithms (or code) such large 'n's.
Processing / Analysis: Alternatively, the tasks involving n-gram frequencies and removing function words or general words through a large text may or will also increase depending on content type.

## Task 2. Text filtering for cleaner feature
1. clean the text used in the task 1; 2. remove all punctuations; 3. convert all characters to their lowercase; 4. remove all words in "stopwords"; 5. remove all relatively meaningless words like " 've ", " 's ", etc. 6. after finishing the above operations, apply stemming and lemmatization to the cleaned text respectively.

```python
# write your code here
import string
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize

# Download required NLTK data
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('wordnet')

# Text from the previous task
train_text = """My wife took me here on my birthday for breakfast and it was excellent.  The weather was perfect which made sitting outside overlooking their grounds an absolute pleasure.
Our waitress was excellent and our food arrived quickly on the semi-busy Saturday morning.  It looked like the place fills up pretty quickly so the earlier you get here the better.

Do yourself a favor and get their Bloody Mary.  It was phenomenal and simply the best I\'ve ever had.
I'm pretty sure they only use ingredients from their garden and blend them fresh when you order it.  It was amazing.

While EVERYTHING on the menu looks excellent, I had the white truffle scrambled eggs vegetable skillet and it was tasty and delicious.
It came with 2 pieces of their griddled bread with was amazing and it absolutely made the meal complete.
It was the best "toast" I\'ve ever had.

Anyway, I can\'t wait to go back!"""

# 1. Remove punctuation
def remove_punctuation(text):
    return text.translate(str.maketrans('', '', string.punctuation))

# 2. Convert to lowercase
def to_lowercase(text):
    return text.lower()

# 3. Remove stopwords
def remove_stopwords(tokens):
    stop_words = set(stopwords.words('english'))
    return [word for word in tokens if word not in stop_words]

# 4. Remove meaningless words
def remove_meaningless_words(tokens):
    meaningless_words = {"'ve", "'s", "'m", "n't", "'re", "'ll"}  # Add more as needed
    return [word for word in tokens if word not in meaningless_words]

# 5. Stemming and Lemmatization
def apply_stemming(tokens):
    stemmer = PorterStemmer()
    return [stemmer.stem(word) for word in tokens]

def apply_lemmatization(tokens):
    lemmatizer = WordNetLemmatizer()
    return [lemmatizer.lemmatize(word) for word in tokens]

# Clean the text
cleaned_text = remove_punctuation(train_text)  # Step 1
cleaned_text = to_lowercase(cleaned_text)  # Step 2
tokens = word_tokenize(cleaned_text)  # Tokenize the text
tokens = remove_stopwords(tokens)  # Step 3
tokens = remove_meaningless_words(tokens)  # Step 4

# Apply stemming
stemmed_tokens = apply_stemming(tokens)

# Apply lemmatization
lemmatized_tokens = apply_lemmatization(tokens)

# Print the results
print("Cleaned and Stemmed Tokens:")
print(stemmed_tokens)
print("\nCleaned and Lemmatized Tokens:")
print(lemmatized_tokens)
```

## Task 3. Apply spacy and textblob chunking to the text used in tesk 1 respectively, and output the noun phrase chunking results

```python
import spacy
from textblob import TextBlob

# Load spaCy's English language model
nlp = spacy.load("en_core_web_sm")

# Text from Task 1
train_text = """My wife took me here on my birthday for breakfast and it was excellent.  The weather was perfect which made sitting outside overlooking their grounds an absolute pleasure.
Our waitress was excellent and our food arrived quickly on the semi-busy Saturday morning.  It looked like the place fills up pretty quickly so the earlier you get here the better.

Do yourself a favor and get their Bloody Mary.  It was phenomenal and simply the best I\'ve ever had.
I'm pretty sure they only use ingredients from their garden and blend them fresh when you order it.  It was amazing.

While EVERYTHING on the menu looks excellent, I had the white truffle scrambled eggs vegetable skillet and it was tasty and delicious.
It came with 2 pieces of their griddled bread with was amazing and it absolutely made the meal complete.
It was the best "toast" I\'ve ever had.

Anyway, I can\'t wait to go back!"""

# 1. Applying spaCy for noun phrase chunking
def spacy_noun_chunks(text):
    doc = nlp(text)
    return [chunk.text for chunk in doc.noun_chunks]

# 2. Applying TextBlob for noun phrase chunking
def textblob_noun_chunks(text):
    blob = TextBlob(text)
    return list(blob.noun_phrases)

# Get noun phrases using spaCy
spacy_chunks = spacy_noun_chunks(train_text)

# Get noun phrases using TextBlob
textblob_chunks = textblob_noun_chunks(train_text)

# Print the results
print("spaCy Noun Phrases:")
print(spacy_chunks)

print("\nTextBlob Noun Phrases:")
print(textblob_chunks)
```

## Question 4. Whats the disadvantage in bag of words  . Please explain in your own words with an example .

### Answer to Q4: The most important limitation of the Bag of Words (BoW) model is it does not handle word order and thus preserves context of word meaning. In the BoW model, one can use the text as only words and ignore the grammar, sentence structure, and dependencies. This is especially true when the order of words in the text has a special meaning, sentiment or intent.

Consider the following two sentences:

"This movie is good, not bad."
"This movie is bad, not good."
From a simple Bag of Words perspective, both sentences have the identical vocabulary of words (and identical words are "This", "movie", "is", "good", "not", and "bad"). Both sentences may also have the same frequency of each of the words as well. But their meaning could not be any different because of the opposite word order that changed the meaning, sentiment, and/or intent. The essential aspect of difference in sentiment is lost due to modelling in a BoW framework.

The BoW model has another weakness with producing a very high dimension feature space, which potentially leads to considerations in practice, especially where large vocabularies are concerned. Each additional word in your corpus is an additional feature, which can give rise to a very wide matrix and therefore, costly computations and potentially leading to the "curse of
