## Task 1
Write a function that will take data and sampling_rate from loaded wav files and return a concatenated feature vector that has all the features from the feature extraction example

```python
#code here
def extract_features(y, sr):
    # Calculate spectral centroid
    cent = librosa.feature.spectral_centroid(y=y, sr=sr)
    # Calculate spectral bandwidth
    spec_bw = librosa.feature.spectral_bandwidth(y=y, sr=sr)
    # Calculate spectral flatness
    flatness = librosa.feature.spectral_flatness(y=y)
    # Calculate spectral rolloff
    rolloff = librosa.feature.spectral_rolloff(y=y, sr=sr)
    # Calculate RMS energy
    rms = librosa.feature.rms(y=y)
    # Calculate zero crossing rate
    zcr = librosa.feature.zero_crossing_rate(y)

    # Compute the mean of each feature
    mean_cent = np.mean(cent)
    mean_spec_bw = np.mean(spec_bw)
    mean_flatness = np.mean(flatness)
    mean_rolloff = np.mean(rolloff)
    mean_rms = np.mean(rms)
    mean_zcr = np.mean(zcr)

    # Concatenate the mean feature values into a single NumPy array
    features = np.array([
        mean_cent,
        mean_spec_bw,
        mean_flatness,
        mean_rolloff,
        mean_rms,
        mean_zcr
    ])

    return features

print("The 'extract_features' function has been defined.")
```

```python
feature_vector = extract_features(y, sr)
print(f"Shape of the extracted feature vector: {feature_vector.shape}")
display(feature_vector)
```

```python
#code here
#code here
import numpy as np
import librosa

# Initialize datasets
Xtrain = []
Ytrain = []
Xtest = []
Ytest = []

# Define sr once, assuming all audio files have the same sampling rate
# The original notebook defines sr from '/content/mnist digit/0_01_0.wav'
# This ensures sr is available for feature extraction
_, sr = librosa.load("/content/mnist digit/0_01_0.wav", sr=None)

# Load training data
for i in range(10):
    for j in [3, 6, 10, 15, 28]:
        x, _ = librosa.load(f"/content/mnist digit/{i}_01_{j}.wav", sr=sr)
        Xtrain.append(x)
        Ytrain.append(i)
        x, _ = librosa.load(f"/content/mnist digit/{i}_36_{j}.wav", sr=sr)
        Xtrain.append(x)
        Ytrain.append(i)

# Load testing data
for i in range(10):
    for j in [12, 23, 31, 35, 42]:
        x, _ = librosa.load(f"/content/mnist digit/{i}_01_{j}.wav", sr=sr)
        Xtest.append(x)
        Ytest.append(i)
        x, _ = librosa.load(f"/content/mnist digit/{i}_36_{j}.wav", sr=sr)
        Xtest.append(x)
        Ytest.append(i)

# Find the maximum length in both train and test datasets
max_train_length = max(len(x) for x in Xtrain)
max_test_length = max(len(x) for x in Xtest)
max_length = max(max_train_length, max_test_length)

# Pad training data
Xtrain_padded = [np.pad(x, (0, max_length - len(x)), mode='constant') for x in Xtrain]

# Pad testing data
Xtest_padded = [np.pad(x, (0, max_length - len(x)), mode='constant') for x in Xtest]

# Print the results to verify
print(f"Training data padded length: {len(Xtrain_padded[0])}, Test data padded length: {len(Xtest_padded[0])}")
print(f"Max length: {max_length}")
```

## Task 3
Apply the function you have written in task 1 to the datasets you have created on task 2. Afterwards use a SVM to train on your  train data and report the accury and f1 score on your test dataset. You can use the svm from scikit-learn https://scikit-learn.org/1.5/modules/svm.html

```python
#code here
#code here
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, f1_score
import numpy as np

# Apply feature extraction to training and testing data
Xtrain_features = [extract_features(x, sr) for x in Xtrain_padded]
Xtest_features = [extract_features(x, sr) for x in Xtest_padded]

# Convert lists to NumPy arrays
Xtrain_features = np.array(Xtrain_features)
Xtest_features = np.array(Xtest_features)
Ytrain = np.array(Ytrain)
Ytest = np.array(Ytest)

# Split data into training and validation sets
X_train, X_val, y_train, y_val = train_test_split(
    Xtrain_features, Ytrain, test_size=0.2, random_state=42
)

# Initialize and train an SVM classifier
svm = SVC(kernel='linear', C=1)  # You can experiment with different kernels and parameters
svm.fit(X_train, y_train)


# Predict on the test set
y_pred = svm.predict(Xtest_features)


# Evaluate the model
accuracy = accuracy_score(Ytest, y_pred)
f1 = f1_score(Ytest, y_pred, average='weighted') # Use weighted average for multi-class

print(f"Accuracy: {accuracy}")
print(f"F1 Score: {f1}")
```
