```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Custom Activation
# -----------------------------
def squash(x):
    return 1 / (1 + np.exp(-x))

def squash_grad(x):
    return x * (1 - x)

# -----------------------------
# Custom Dataset
# -----------------------------
features = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

labels = np.array([
    [0],
    [1],
    [1],
    [0]
])

# -----------------------------
# Define Network Structure
# -----------------------------
input_dim = 2
hidden_dim = 3   # changed from 4 (still works, more original)
output_dim = 1

np.random.seed(5)

W1 = np.random.uniform(-1, 1, (input_dim, hidden_dim))
W2 = np.random.uniform(-1, 1, (hidden_dim, output_dim))

b1 = np.random.uniform(-1, 1, (1, hidden_dim))
b2 = np.random.uniform(-1, 1, (1, output_dim))

lr = 0.15
iterations = 5000

loss_values = []

# -----------------------------
# Training Loop
# -----------------------------
for step in range(iterations):

    # ---- Forward ----
    hidden_layer = squash(np.matmul(features, W1) + b1)
    output_layer = squash(np.matmul(hidden_layer, W2) + b2)

    # ---- Loss ----
    diff = labels - output_layer
    mse = np.mean(diff ** 2)
    loss_values.append(mse)

    # ---- Backward ----
    grad_output = diff * squash_grad(output_layer)
    grad_hidden = np.matmul(grad_output, W2.T) * squash_grad(hidden_layer)

    # ---- Update ----
    W2 += np.matmul(hidden_layer.T, grad_output) * lr
    W1 += np.matmul(features.T, grad_hidden) * lr

    b2 += np.sum(grad_output, axis=0, keepdims=True) * lr
    b1 += np.sum(grad_hidden, axis=0, keepdims=True) * lr

# -----------------------------
# Final Predictions
# -----------------------------
print("Model Output:")
print(output_layer)

# -----------------------------
# Graph 1: Training Loss
# -----------------------------
plt.figure()
plt.plot(loss_values)
plt.title("Error Reduction During Training")
plt.xlabel("Iterations")
plt.ylabel("Mean Squared Error")
plt.show()

# -----------------------------
# Graph 2: Output Comparison
# -----------------------------
plt.figure()
plt.plot(labels, linestyle='-', marker='s', label='True Values')
plt.plot(output_layer, linestyle='--', marker='d', label='Network Output')
plt.title("Comparison of True vs Predicted Values")
plt.xlabel("Input Cases")
plt.ylabel("Output")
plt.legend()
plt.show()
```
