```python
import random
import math
import matplotlib.pyplot as plt

# Step 1: Dataset (slightly modified for uniqueness)
dataset_values = [(1,2), (2,1), (3,4), (5,7), (3,3), (8,7), (9,8), (10,10), (6,5)]

# Step 2: Parameters
num_clusters = 2
max_iters = 100

# Step 3: Distance function
def compute_distance(a, b):
    return math.sqrt((a[0] - b[0])**2 + (a[1] - b[1])**2)

# Step 4: Initialize centroids randomly
centers = random.sample(dataset_values, num_clusters)

for _ in range(max_iters):
    grouped_points = [[] for _ in range(num_clusters)]

    # Assignment step
    for point in dataset_values:
        dist_list = [compute_distance(point, c) for c in centers]
        index = dist_list.index(min(dist_list))
        grouped_points[index].append(point)

    # Update step
    updated_centers = []
    for grp in grouped_points:
        if len(grp) > 0:
            avg_x = sum(p[0] for p in grp) / len(grp)
            avg_y = sum(p[1] for p in grp) / len(grp)
            updated_centers.append((avg_x, avg_y))
        else:
            updated_centers.append(random.choice(dataset_values))

    # Convergence check
    if updated_centers == centers:
        break

    centers = updated_centers

# Step 5: Compute WCSS (metric)
wcss = 0
for i in range(num_clusters):
    for point in grouped_points[i]:
        wcss += compute_distance(point, centers[i])**2

# Step 6: Output
print("Final Centroids:", centers)
print("Clusters:", grouped_points)
print("WCSS (Inertia):", wcss)

# Step 7: Visualization
for i in range(num_clusters):
    x_vals = [p[0] for p in grouped_points[i]]
    y_vals = [p[1] for p in grouped_points[i]]
    plt.scatter(x_vals, y_vals, label=f"Cluster {i+1}")

# Plot centroids
centroid_x = [c[0] for c in centers]
centroid_y = [c[1] for c in centers]
plt.scatter(centroid_x, centroid_y, marker='X', s=200, label='Centroids')

plt.title("K-Means Clustering Result")
plt.legend()
plt.show()
```
