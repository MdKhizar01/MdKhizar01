```python
import random

# Convert binary string to integer
def decode(chromosome):
    return int(chromosome, 2)

# Fitness function (same mathematical goal, but used independently)
def evaluate(chromosome):
    x = decode(chromosome)
    return (x * x) + (2 * x) + 1

# Create a random chromosome
def create_chromosome(length):
    return ''.join(random.choice(['0', '1']) for _ in range(length))

# Generate initial population
def create_population(pop_size, chrom_length):
    return [create_chromosome(chrom_length) for _ in range(pop_size)]

# Rank-based selection (alternative to roulette wheel)
def rank_selection(population):
    sorted_pop = sorted(population, key=lambda c: evaluate(c), reverse=True)
    weights = list(range(1, len(sorted_pop) + 1))
    return random.choices(sorted_pop, weights=weights, k=2)

# One-point crossover
def recombine(p1, p2):
    cut = random.randint(1, len(p1) - 1)
    offspring1 = p1[:cut] + p2[cut:]
    offspring2 = p2[:cut] + p1[cut:]
    return offspring1, offspring2

# Bit-flip mutation
def apply_mutation(chromosome, rate=0.1):
    mutated = list(chromosome)
    for i in range(len(mutated)):
        if random.random() < rate:
            mutated[i] = '1' if mutated[i] == '0' else '0'
    return ''.join(mutated)

# Main GA loop
def run_ga():
    POP_SIZE = 6
    CHROM_LENGTH = 5
    GENERATIONS = 5
    MUTATION_RATE = 0.1

    population = create_population(POP_SIZE, CHROM_LENGTH)

    print("Initial Population:")
    for c in population:
        print(c, decode(c), evaluate(c))

    for g in range(GENERATIONS):
        next_gen = []

        while len(next_gen) < POP_SIZE:
            parent1, parent2 = rank_selection(population)

            child1, child2 = recombine(parent1, parent2)

            child1 = apply_mutation(child1, MUTATION_RATE)
            child2 = apply_mutation(child2, MUTATION_RATE)

            next_gen.append(child1)
            if len(next_gen) < POP_SIZE:
                next_gen.append(child2)

        population = next_gen

        print(f"\nGeneration {g+1}:")
        for c in population:
            print(c, decode(c), evaluate(c))

    return population

# Execute GA
run_ga()
```

```python
import random

class GeneticAlgorithm:
    def __init__(self, pop_size=8, chrom_length=6, generations=6, mutation_rate=0.2):
        self.pop_size = pop_size
        self.chrom_length = chrom_length
        self.generations = generations
        self.mutation_rate = mutation_rate
        self.population = []

    # Convert binary to integer
    def decode(self, chromosome):
        return int(chromosome, 2)

    # Fitness function
    def evaluate(self, chromosome):
        x = self.decode(chromosome)
        return (x + 1) ** 2 + 10

    # Create one chromosome
    def create_individual(self):
        return ''.join(random.choice(['0', '1']) for _ in range(self.chrom_length))

    # Initialize population
    def seed_population(self):
        self.population = [self.create_individual() for _ in range(self.pop_size)]

    # Selection using weighted probabilities
    def select_parent(self):
        weights = [self.evaluate(ind) for ind in self.population]
        total = sum(weights)
        probabilities = [w / total for w in weights]
        return random.choices(self.population, weights=probabilities, k=1)[0]

    # Uniform crossover
    def crossover(self, parent_a, parent_b):
        child_a = []
        child_b = []

        for i in range(len(parent_a)):
            if random.random() < 0.5:
                child_a.append(parent_a[i])
                child_b.append(parent_b[i])
            else:
                child_a.append(parent_b[i])
                child_b.append(parent_a[i])

        return ''.join(child_a), ''.join(child_b)

    # Mutation operation
    def mutate(self, chromosome):
        mutated = list(chromosome)
        for i in range(len(mutated)):
            if random.random() < self.mutation_rate:
                mutated[i] = '1' if mutated[i] == '0' else '0'
        return ''.join(mutated)

    # Run one generation
    def evolve(self):
        next_gen = []

        while len(next_gen) < self.pop_size:
            p1 = self.select_parent()
            p2 = self.select_parent()

            c1, c2 = self.crossover(p1, p2)

            c1 = self.mutate(c1)
            c2 = self.mutate(c2)

            next_gen.append(c1)
            if len(next_gen) < self.pop_size:
                next_gen.append(c2)

        self.population = next_gen[:self.pop_size]

    # Run full GA
    def run(self):
        self.seed_population()

        print("Initial Population:")
        for ind in self.population:
            print(ind, self.decode(ind), self.evaluate(ind))

        for gen in range(self.generations):
            self.evolve()

            print(f"\nGeneration {gen+1}:")
            for ind in self.population:
                print(ind, self.decode(ind), self.evaluate(ind))


# Execute GA
ga = GeneticAlgorithm()
ga.run()
```
