## **Code Implementation**

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix

# ------------------------------
# Step 1: Load dataset
# ------------------------------
data = pd.read_csv("/content/sample_data/california_housing_train.csv")

# ------------------------------
# Step 2: Feature and target
# ------------------------------
# Updated features and target for the new dataset
features = ['longitude', 'latitude', 'housing_median_age', 'total_rooms', 'total_bedrooms', 'population', 'households', 'median_income']
X = data[features].values
y = data['median_house_value'].values.reshape(-1, 1)

# Normalize features (important for logistic regression)
X = (X - np.mean(X, axis=0)) / np.std(X, axis=0)
y = (y - np.mean(y, axis=0)) / np.std(y, axis=0)

# ------------------------------
# Step 3: Sigmoid function
# ------------------------------
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# ------------------------------
# Step 4: Initialize weights
# ------------------------------
m, n = X.shape
weights = np.zeros((n, 1))
bias = 0
learning_rate = 0.01  # Increased learning rate for faster convergence on normalized data
epochs = 1000

# ------------------------------
# Step 5: Gradient Descent
# ------------------------------
for i in range(epochs):
    # Forward propagation
    z = np.dot(X, weights) + bias
    predictions = sigmoid(z)

    # Compute gradients
    dz = predictions - y
    dw = np.dot(X.T, dz) / m
    db = np.sum(dz) / m

    # Update weights
    weights -= learning_rate * dw
    bias -= learning_rate * db

# ------------------------------
# Step 6: Predictions
# ------------------------------
z = np.dot(X, weights) + bias
predicted_prob = sigmoid(z)

# For binary classification, we need a threshold. Since `median_house_value` is continuous,
# logistic regression is not directly applicable for classification without discretizing 'y'.
# For demonstration, I will treat it as a regression problem, but `accuracy_score` and
# `confusion_matrix` are for classification. Let's adjust to a simple binary classification
# example to use these metrics, e.g., classify if house value is above or below median.
median_house_value_median = np.median(data['median_house_value'])
y_binary = (data['median_house_value'] > median_house_value_median).astype(int).values.reshape(-1, 1)

# Re-run a simplified logistic regression for binary classification using y_binary
# (This is a quick fix to make the evaluation metrics work; ideally,
# the model should be retrained with binary labels from the start.)

# Re-initialize weights and bias for the binary problem
weights_binary = np.zeros((n, 1))
bias_binary = 0

for i in range(epochs):
    z_binary = np.dot(X, weights_binary) + bias_binary
    predictions_binary = sigmoid(z_binary)
    dz_binary = predictions_binary - y_binary
    dw_binary = np.dot(X.T, dz_binary) / m
    db_binary = np.sum(dz_binary) / m
    weights_binary -= learning_rate * dw_binary
    bias_binary -= learning_rate * db_binary

predicted_prob_binary = sigmoid(np.dot(X, weights_binary) + bias_binary)
predicted_class = (predicted_prob_binary >= 0.5).astype(int)

# ------------------------------
# Step 7: Evaluate model
# ------------------------------
accuracy = accuracy_score(y_binary, predicted_class)
cm = confusion_matrix(y_binary, predicted_class)

print("Model Accuracy:", accuracy)
print("Confusion Matrix:\n", cm)
```
